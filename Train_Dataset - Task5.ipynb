## Univariate Analysis

```python
fig, axes = plt.subplots(3, 2, figsize=(14, 12))
fig.suptitle("Univariate Analysis", fontsize=16)

sns.histplot(df['Age'].dropna(), bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Distribution of Age")
axes[0, 0].set_xlabel("Age")
axes[0, 0].set_ylabel("Count")

sns.histplot(df['Fare'], bins=30, kde=True, ax=axes[0, 1])
axes[0, 1].set_title("Distribution of Fare")
axes[0, 1].set_xlabel("Fare")
axes[0, 1].set_ylabel("Count")

sns.countplot(x='Pclass', data=df, ax=axes[1, 0])
axes[1, 0].set_title("Passenger Class Distribution")
axes[1, 0].set_xlabel("Pclass")
axes[1, 0].set_ylabel("Count")

sns.countplot(x='Sex', data=df, ax=axes[1, 1])
axes[1, 1].set_title("Gender Distribution")
axes[1, 1].set_xlabel("Sex")
axes[1, 1].set_ylabel("Count")

sns.boxplot(x='Survived', y='Age', data=df, ax=axes[2, 0])
axes[2, 0].set_title("Age Distribution by Survival")
axes[2, 0].set_xlabel("Survived (0 = No, 1 = Yes)")
axes[2, 0].set_ylabel("Age")

sns.countplot(x='Embarked', data=df, ax=axes[2, 1])
axes[2, 1].set_title("Port of Embarkation")
axes[2, 1].set_xlabel("Embarked")
axes[2, 1].set_ylabel("Count")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

**Observations:**

1)The age distribution is right-skewed. Most passengers are between 20 and 40 years old. There are also many children onboard. There are some missing values in the 'Age' column.

2)Fare is heavily right-skewed, with most passengers paying under 100. A few outliers paid more than 500.

3)Most passengers belonged to the 3rd class, followed by 1st and 2nd classes. Class 3 passengers were likely lower-income.

4)There are significantly more male passengers than female passengers (~65% male).

5)Median age of survivors is slightly lower than that of non-survivors. Young children had a higher survival chance.

6)Most passengers boarded from port 'S' (Southampton), followed by 'C' (Cherbourg) and 'Q' (Queenstown).


## Scatter Plot

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Age", y="Fare", hue="Survived", palette="Set1", alpha=0.7)
plt.title("Age vs Fare (Colored by Survival)")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived", loc='upper right')
plt.show()
```

**Observation:**

- Many high-fare, young/middle-aged passengers survived.
- Dense cluster of low-fare passengers didn’t survive as often.


## Pair Plots

```python
pairplot_cols = ['Survived', 'Pclass', 'Age', 'Fare', 'SibSp', 'Parch']
sns.pairplot(df[pairplot_cols], hue='Survived', palette='Set1', diag_kind='kde')
plt.suptitle("Pairplot of Selected Features", y=1.02)
plt.show()
```

 **Observation:**

- **Survival vs. Pclass**: 
  - Passengers in **1st class** had higher survival rates (many blue points).
  - Most passengers in **3rd class** did not survive (many red points).

- **Survival vs. Age**: 
  - Survivors include a cluster of **younger passengers**, especially children.
  - Older age groups had a lower survival rate overall.

- **Survival vs. Fare**:
  - **Higher fares** are associated with higher survival chances.
  - Many low-fare passengers (likely from 3rd class) didn’t survive.

- **Survival vs. SibSp & Parch**:
  - People with **1–2 siblings/spouses** or **1–3 parents/children** aboard had higher chances of survival.
  - Very high values of SibSp/Parch were rare and associated more with non-survivors.

- **Correlations**:
  - There's a visible **relationship between Pclass and Fare**: 1st class passengers paid higher fares.
  - **Fare and Age** also loosely trend together—older or wealthier passengers paid more.

## Box Plot

```python
df.boxplot(figsize=(20,10))
```

 **Observation:**

- **Age** and **Fare** show significant spread and presence of outliers:
  - **Age**: Most passengers fall in the range of 20–40 years, with outliers extending beyond 60.
  - **Fare**: Highly right-skewed with extreme outliers. A few passengers paid over 500, while most fares are below 100.
- **SibSp**, **Parch**, and **FamilySize** have tight distributions near zero, indicating most passengers were either alone or with very few family members.
- **Pclass** and **Survived** have limited numerical spread since they are categorical in nature (ordinal/binary).
- **PassengerId** is a unique identifier and not useful for statistical analysis, but it's shown here for completeness.


# Summary of Findings

- **Survival Rate** was ~38%.
- **Gender played a critical role** in survival: females had much higher survival rates.
- **Class mattered**: passengers in 1st class had significantly higher survival chances than those in 3rd class.
- **Children and young adults** were more likely to survive.
- **Fare paid** is positively correlated with survival: higher-paying passengers tended to survive.
- **Most passengers boarded at Southampton**, but survival rate varied slightly across ports.
- Some columns like `Cabin` had a lot of missing data, and may need further treatment or exclusion.
